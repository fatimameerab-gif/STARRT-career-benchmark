# STARRT.ai — reviewer checks (DistilBERT)

**Settings:** Accelerator = **GPU T4 x2**, Internet = **On**. *Add Input* → `starrt-review` dataset.

Part A: 2 extra seeds (Djinni + Faruque). Part B: Djinni ablations. Results → `starrt_review_results.zip`.

```python
import glob, os, subprocess, sys
hits = glob.glob('/kaggle/input/**/run_distilbert.py', recursive=True)
assert hits, 'Dataset not attached: Add Input -> starrt-review'
CODE = os.path.dirname(hits[0]); print('code folder:', CODE)
import torch; print('GPU:', torch.cuda.is_available(), torch.cuda.get_device_name(0) if torch.cuda.is_available() else '')

def run(script, args=''):
    cmd = f'cd /kaggle/working && python -u {CODE}/{script} {args}'
    print('>>', cmd, flush=True); r = subprocess.run(cmd, shell=True); assert r.returncode == 0, f'{script} failed'
```

## 0. Smoke test (2 min)

```python
run('run_distilbert.py', '--datasets djinni_masked --folds 0 --epochs 1 --limit 400 --skip_forms --tag smoke')
print('smoke test OK')
```

## A. Extra seeds, Djinni + Faruque (~1.5 h)

```python
for seed in [13, 7]:
    run('run_distilbert.py', f'--datasets djinni,faruque --seed {seed} --tag distilbert_seed{seed} --skip_forms')
```

## B. Djinni ablations, seed 42 (~1.5 h)

```python
run('run_distilbert.py', '--datasets djinni_nolook,djinni_masked,djinni_tools --tag distilbert --skip_forms')
```

## Zip results
Right side **Output** panel se `starrt_review_results.zip` download kar ke chat mein upload karein.

```python
import shutil
for f in glob.glob('/kaggle/working/results/smoke*'): os.remove(f)
shutil.make_archive('/kaggle/working/starrt_review_results', 'zip', '/kaggle/working/results')
print(sorted(os.listdir('/kaggle/working/results')))
```
